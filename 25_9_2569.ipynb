```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.random.rand(50)
Y = 3 * X + 1 + 0.5 * np.random.rand(50)
plt.plot(X, Y, '.r')
plt.show()
```

```python
X_ = np.vstack((X, np.ones(len(X))))
```

```python
R = Y @ X_.T @ np.linalg.inv(X_ @ X_.T)
```
